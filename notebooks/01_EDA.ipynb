# House Price Prediction — Exploratory Data Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_style("whitegrid")
```

```python
df = pd.read_csv("../Housing.csv")
print(f"Shape: {df.shape}")
df.head()
```

```python
df.describe()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0, 0].hist(df["price"] / 100000, bins=50, edgecolor="black")
axes[0, 0].set_title("Price Distribution (in lakhs)")
axes[0, 0].set_xlabel("Price (Rs. Lakhs)")

axes[0, 1].scatter(df["area"], df["price"] / 100000, alpha=0.3)
axes[0, 1].set_title("Price vs Area")
axes[0, 1].set_xlabel("Area (sqft)")
axes[0, 1].set_ylabel("Price (Rs. Lakhs)")

axes[1, 0].scatter(df["bedrooms"], df["price"] / 100000, alpha=0.3)
axes[1, 0].set_title("Price vs Bedrooms")
axes[1, 0].set_xlabel("Bedrooms")
axes[1, 0].set_ylabel("Price (Rs. Lakhs)")

axes[1, 1].scatter(df["parking"], df["price"] / 100000, alpha=0.3)
axes[1, 1].set_title("Price vs Parking")
axes[1, 1].set_xlabel("Parking spots")
axes[1, 1].set_ylabel("Price (Rs. Lakhs)")

plt.tight_layout()
plt.show()
```

```python
corr = df.select_dtypes(include=[np.number]).corr()["price"].sort_values(ascending=False)
plt.figure(figsize=(10, 6))
corr.drop("price").plot(kind="bar")
plt.title("Correlation with Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
